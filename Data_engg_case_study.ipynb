```python
tree = ET.parse('/content/DATA/literacy_rate.xml')
root = tree.getroot()
```

```python
#TASK 4

rows = []

for country in root.findall('country'):

    name = country.get('name')
    rate = country.find('literacy_rate').text

    rows.append({
        'country': name,
        'literacy_rate': rate
    })
```

```python
literacy = pd.DataFrame(rows)

literacy.head()
```

```python
#TASK 6
gdp.rename(columns={'Country': 'country'}, inplace=True)
```

```python
print(population.columns)
print(gdp.columns)
print(internet.columns)
print(literacy.columns)
```

```python
population['population'] = pd.to_numeric(
    population['population'], errors='coerce'
)

gdp['GDP'] = pd.to_numeric(
    gdp['GDP'], errors='coerce'
)

internet['internet_users'] = pd.to_numeric(
    internet['internet_users'], errors='coerce'
)

literacy['literacy_rate'] = pd.to_numeric(
    literacy['literacy_rate'], errors='coerce'
)
```

```python
print(population.dtypes)
print(gdp.dtypes)
print(internet.dtypes)
print(literacy.dtypes)
```

```python
merged = pd.merge(
    population, gdp,
    on='country',
    how='outer'
)

merged = pd.merge(
    merged, internet,
    on='country',
    how='outer'
)

merged = pd.merge(
    merged, literacy,
    on='country',
    how='outer'
)

merged.head()
```

```python
merged.shape
```

```python
merged.isnull().sum()
```

```python
merged['Internet Penetration (%)'] = (
    merged['internet_users'] / merged['population']
) * 100

merged['Internet Penetration (%)'] = (
    merged['Internet Penetration (%)'].round(2)
)
```

```python
merged['GDP Per Capita'] = (
    merged['GDP'] / merged['population']
).round(2)
```

```python
merged.head()
```

```python
#PART D
import sqlite3

conn = sqlite3.connect('country_statistics.db')

merged.to_sql(
    'country_statistics',
    conn,
    if_exists='replace',
    index=False
)
```

```python
#TASK 11
query = """
SELECT *
FROM country_statistics
LIMIT 5;
"""

pd.read_sql_query(query, conn)
```
